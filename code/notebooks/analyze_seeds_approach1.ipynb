```python
import json, glob
import pandas as pd

rows = []
for path in sorted(glob.glob("../runs/gbert_seed*/test_results.json")):
    m = json.load(open(path))
    m["run"] = path.split("/")[-2]
    rows.append(m)

res = pd.DataFrame(rows).set_index("run")
cols = ["test_f1_macro", "test_f1_hate", "test_f1_not_hate", "test_precision_hate", "test_recall_hate"]
display(res[cols].round(4))
display(res[cols].agg(["mean", "std"]).round(4))
```

```python
summary = res[cols].agg(["mean", "std"]).T
summary["report"] = summary.apply(lambda r: f"{100*r['mean']:.1f} ± {100*r['std']:.1f}", axis=1)
display(summary["report"])
```
